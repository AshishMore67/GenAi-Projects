```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

### **Call LLM Using Google API key**

```python
llm = ChatGoogleGenerativeAI(model="gemma-4-31b-it")

response = llm.invoke("How many moons does Jupiter have?")
print(response.text)
```

```python
llm = ChatGoogleGenerativeAI(model="gemma-4-31b-it")

response = llm.invoke([
    ["system", "You are a helpful assistant that answers in short one line"],
    ["human", "tell me about yourself"]
])
print(response.text)
```

### Temprature

```python
prompt = "Give me top 5 buisness ideas in logistics domain. Only ideas no explainations"

llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash",temperature=0)
response = llm.invoke(prompt)
print(response.text)
```

```python
prompt = "Give me top 5 buisness ideas in logistics domain.Only ideas no explainations"

llm = ChatGoogleGenerativeAI(model="gemini-3.6-flash",temperature=1)
response = llm.invoke(prompt)
print(response.text)
```

### **Call LLM using Groq Models**

```python
from langchain_groq import ChatGroq
```

```python
llm = ChatGroq(model="openai/gpt-oss-20b", reasoning_format="parsed")

messages = [
    (
        "system",
        "You are a helpful assistant that gives a short answer to a question",
    ),
    (
        "human", 
        "How many wordcups won by Brazil and when?"
    ),
]
ai_msg = llm.invoke(messages)
print(ai_msg.content)
```
